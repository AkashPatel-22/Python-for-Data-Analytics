```python
# Data Types
# - A Data Types defines the 'kind of values a varible can hold'.
```

```python
int # integer values.
float # decimal values.
str # string- squence of character.
bool # boolean values - true/false.
None # no value/null.
```

```python
# type() - use to check type of datatypes
name = "Aakash Patel"
age = 23
PI = 3.14
isTeacher = True
empty = None

print(type(name))
print(type(age))
print(type(PI))
print(type(isTeacher))
print(type(empty))
```

```python
# Type Conversion & Type Casting.
# - when we convert variable from one type to another

# 1.Type Conversion - implicit - done automatically by python.
a = 5
b = 3.0
print(a+b)
# python convert ans in float by default.
```

```python
# 2. Type Casting.
x = 10  #int
y = float(x)  #convert to float
z = str(x)  #convert to string

print("values are:",x,y,z)
print(type(x))
print(type(y))
print(type(z))
```
