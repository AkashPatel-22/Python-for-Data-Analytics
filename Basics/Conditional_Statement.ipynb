```python
# there are 3 main conditional statements python:
# 1. if - used to test a condition. if its true, the indented block runs.
# 2. else - else block runs if all above conditions are false.
# 3. elif - ("else if") used when we have multiple conditions to check.
```

## if-elif-else Conditional

```python
# Example- 1
age = int(input("enter age: "))

if age >= 18:
    print("you can vote") 
    print("you can drive")
else:
    print("you can't vote")
    print("you can't dive")
```

```python
# Example 2 - Traffic Lights
color = input("enter color: ")

if color == "red":
    print("STOP")
elif color == "yellow":
    print("WAIT")
elif color == "gren":
    print("GO")
else:
    print("wrong color")
```

```python
# multiple conditions using and or , not operator.

# example 3 - logical operators in expression.
age = int(input("enter age: "))

if age < 13:
    print("child")
elif(age>=13 and age<18):
    print("teenager")
else:
    print("adult")
```

## Nesting in Conditionals

```python
# Nesting means placing one block of code inside another.
# it means putting one 'if' or 'if-elif-else' inside another.
# this helps when you need to make a decision based on another decision.
```

```python
# example -login system.

username = input("enter user name: ")
password = input("enter password")

if username == "admin" and password == "pass":
    print("Login Succesfully")
else:
    if username != "admin":
            print("wrong username, try again")
    else:
            print("wrong password, try again")
```

## Ternary Statements

```python
# -Ternary Statements (or conditional expression) are just another comapct way to writing our if-else statements in one-line
```

```python
age = 18
status = 'Adult' if age >= 18 else "Not Adult"
print(status)
```

## Match Case Statements

```python
# Match / case statements are an alternative to long chain of if..elif..else statements.
```

```python
# # Match case Sytax
# match variable:
#     case pattern1:
#         # code...
#     case  pattern2:
#         #  code...
#     case _:
#         # default
```

```python
# 1. Match - what you're comparing
# 2. case - the value or patterns to match against
# 3. _ - wildcard (matches anything, like "default")
```

# example.
color = input("enter color: ")
match color:
    case "red":
        print("STOP")
    case "yellow":
        print("LOOk")
    case "green":
        print("GO")
    case _:
        print("wrong color")
