```python
# QS - program that takes input of 2 numbers (a & b) and print the average.

a = int(input("enter 1st num: ")) # 5
b = int(input("enter 2nd num: ")) # 10

avg = (a+b)/2

print("average = ", avg)  # 7.5
```
