```python
# operator
# - it is a symbol that tells python to perform a specific computation on one or more values(called operands)

print(1+3) # '+' operator & '1,3' operands

# 4-Differents types of operator in python.
```

```python
# 1. Arithmetic Operators - used to perform math operations.
# operators- +,-,*,/,%,**

a = 5
b = 10

print(5+10) # addition
print(a+b)

print(a-b)  # subtraction
print(a*b)  # multiplication
print(a/b)  # division
print(a%b)  # modulo - remainder
print(a**b) # power
```

```python
# 2. Relational Operators - used to compare values.
# operators - ==,!=, >=, <=, <, > 

a = 10
b = 20

# equal to
print(a == b) # False

# not equal to
print(a!=b) #true

# greater than
print(a>b)  #false

# less than 
print(a<b)  # true

# greater than or equal to
print(a>=b) # false

# less than or equal to
print(a<=b) # true
```
